## A simple bar chart

[Ben Alex Keen](http://benalexkeen.com/bar-charts-in-matplotlib/) tells us that bar charts "are used to display values associated with categorical data." Someone at [Yale University](http://www.stat.yale.edu/Courses/1997-98/101/catdat.htm) tells us that "Categorical variables represent types of data which may be divided into groups. Examples of categorical variables are race, sex, age group, and educational level."

Bar charts are also often used to provide visual comparisons of such values.

Assume we want to compare the heights in inches of the members of a basketball team. We can do that with the following code that uses the [Axes.bar](https://matplotlib.org/api/_as_gen/matplotlib.axes.Axes.bar.html?highlight=bar#matplotlib.axes.Axes.bar) method to create a bar chart showing the height of each player identified by the player's name. Note that the horizontal arrangement of the bars in a bar chart is often not critical. In this case, we will arrange the bars alphabetically by the player's name.

```python
names = ['Alex','Bob','Don','Frank','Harry']
xPosition = range(1,6)
heights = [78,62,72,60,75]

fig,ax = plt.subplots(1,1)
ax.bar(xPosition,heights,tick_label=names)

ax.grid()
fig.tight_layout()
plt.savefig("VisualizationPart07.zzz.png")
plt.show()
```

### Overall Summary

### Detailed Description

### Purpose of the Image
The purpose of this image is to demonstrate how a bar chart can be used to display values associated with categorical data. In this case, the categories are player names, and the values are their heights. The chart illustrates how bar charts provide a clear visual comparison of numerical values across different categories.

## Cosmetic improvement

The following code adds some cosmetic improvements to the bar chart by adding labels to the axes and adding a title.

```python
names = ['Alex','Bob','Don','Frank','Harry']
xPosition = range(1,6)
heights = [78,62,72,60,75]

fig,ax = plt.subplots(1,1)
ax.bar(xPosition,heights,tick_label=names)
ax.set_xlabel('Player Name')
ax.set_ylabel('Player Height in Inches')
ax.set_title('Player Height Comparison')

ax.grid()
fig.tight_layout()
plt.savefig("VisualizationPart07.zzz.png")
plt.show()
```

## Bar chart with multiple x-values

Suppose we want to compare the individual scores achieved by each of our five basketball players in three consecutive games. We can accomplish that with the following code, which shows the three scores for each player in a side-by-side bar chart display. This chart also includes a legend that links the colors to the game number.

```python
names = ['Alex','Bob','Don','Frank','Harry']

xPosition = np.arange(5)
game01 = [78,62,72,60,75]
game02 = [34,70,28,54,36]
game03 = [48,50,60,46,62]

width = 0.25

fig,ax = plt.subplots(1,1)
ax.bar(xPosition,game01,width,label="Game 1")
ax.bar(xPosition+width,game02,width,label="Game 2",tick_label=names)
ax.bar(xPosition+width*2,game03,width,label="Game 3")
ax.set_xlabel('Player Name')
ax.set_ylabel('Player Scores')
ax.set_title('Player Score Comparison')

ax.grid()
plt.legend(loc='lower left')
fig.tight_layout()
plt.savefig("VisualizationPart07.zzz.png")
plt.show()
```

### Overall Summary

### Purpose of the Image

In this case, the categories are player names, and the values are the scores achieved in three different games. The chart illustrates how side‑by‑side bars allow viewers to compare performance across games for each player and to identify patterns or differences among players.

## A stacked bar chart

Another way to compare the individual scores achieved by each of our five players in three consecutive games is to stack the scores. We can accomplish that with the following code.

```python
names = ['Alex','Bob','Don','Frank','Harry']

xPosition = np.arange(5)
game01 = np.array([78,62,72,60,75])
game02 = np.array([34,70,28,54,36])
game03 = np.array([48,50,60,46,62])

fig,ax = plt.subplots(1,1)
ax.bar(xPosition,game01,bottom=0,label="Game 1",tick_label=names)
ax.bar(xPosition,game02,bottom=game01,label="Game 2")
ax.bar(xPosition,game03,bottom=(game01+game02),label="Game 3")
ax.set_xlabel('Player Name')
ax.set_ylabel('Player Scores')
ax.set_title('Player Score Comparison')

ax.grid()
plt.legend(loc='lower left')
fig.tight_layout()
plt.savefig("VisualizationPart07.zzz.png")
plt.show()
```

### Type of Image
This figure is a stacked vertical bar chart. For each player, three colored segments are stacked on top of one another to show the scores achieved in three different games. A legend identifies the game associated with each color.

### Overall Summary
The image provides a comparison of total and component scores for five players across three consecutive games. Each bar represents one player, and the stacked segments show how much each game contributes to the player's overall score. The chart makes it easy to compare both individual game performance and total performance among the players.

### Detailed Description
The horizontal axis is labeled “Player Name” and displays five names: Alex, Bob, Don, Frank, and Harry. Above each name is a single vertical bar composed of three stacked segments. The vertical axis is labeled “Player Scores” and shows numerical values increasing from 0 to 175 in increments of 25. The bottom segment of each bar represents Game 1, the middle segment represents Game 2, and the top segment represents Game 3. Bob’s bar is the tallest, indicating the highest total score. The other players have similar total heights, with variations in how the individual game scores contribute to the total. A legend in the lower left corner identifies the colors associated with Game 1, Game 2, and Game 3. A grid is visible behind the bars, making it easier to estimate the height of each segment. The title “Player Score Comparison” appears above the chart.

### Purpose of the Image
The purpose of this image is to demonstrate how a stacked bar chart can be used to compare multiple values within each category while also showing the total value. In this case, the categories are player names, and the values are the scores achieved in three different games. The chart illustrates how stacked bars allow viewers to compare both individual game contributions and overall performance among players.

## A basic pie chart

[Wikipedia](https://en.wikipedia.org/wiki/Pie_chart) describes a pie chart as follows:

*"A Pie chart (or a circle chart) is a circular statistical graphic which is divided into slices to illustrate numerical proportion. In a pie chart, the arc length of each slice (and consequently its central angle and area), is proportional to the quantity it represents. While it is named for its resemblance to a pie which has been sliced, there are variations on the way it can be presented. The earliest known pie chart is generally credited to William Playfair's Statistical Breviary of 1801.*

*Pie charts are very widely used in the business world and the mass media. However, they have been criticized and many experts recommend avoiding them, pointing out that research has shown it is difficult to compare different sections of a given pie chart, or to compare data across different pie charts. Pie charts can be replaced in most cases by other plots such as the bar chart, box plot or dot plots."*

If you examine the [pie chart documentation](https://matplotlib.org/2.0.2/api/_as_gen/matplotlib.axes.Axes.pie.html#matplotlib.axes.Axes.pie) you will see that the **Axes.pie** method takes many parameters that can be used to customize the appearance of the chart. 

Continuing with our basketball theme, we will create a pie chart that can be used to visually compare the number of points scored by the player named Alex in seven consecutive games.

```python
games = 'Game 1', 'Game 2', 'Game 3', 'Game 4', 'Game 5','Game 6','Game 7'
points = [78, 34, 48, 20, 62, 37, 29]# points scored during each game
explodeSlices = (0.2, 0, 0, 0.3, 0, 0, 0)#explode largest and smallest
percentages = '%1.1f%%' #show percentage of each slice

fig1, ax = plt.subplots()
ax.pie(points, explode=explodeSlices, labels=games, autopct=percentages,
        shadow=True, startangle=90)
ax.axis('equal')  # ensure that pie is drawn as a circle.
ax.set_title('Comparison of scores by game for Alex')
fig.tight_layout()
plt.savefig("VisualizationPart07.zzz.png")
plt.show()
```
